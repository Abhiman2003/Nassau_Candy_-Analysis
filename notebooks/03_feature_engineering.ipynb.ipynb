```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/Nassau Candy Distributor.csv")
```

```python
print(df.shape)
df.head()
```

```python
print(df.columns.tolist())
```

```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    dayfirst=True
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    dayfirst=True
)
```

```python
print(df[["Order Date", "Ship Date"]].head(10))
```

```python
df["Shipping Lead Time"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days
```

```python
print("Shipping Lead Time:", df["Shipping Lead Time"].head(20))
```

```python
print(df["Ship Mode"].value_counts())
```

```python
ship_mode_analysis = (
    df.groupby("Ship Mode")["Shipping Lead Time"]
      .agg(["count", "mean", "min", "max"])
      .round(2)
)

print(ship_mode_analysis)
```

```python
ship_mode_variability = (
    df.groupby("Ship Mode")["Shipping Lead Time"]
      .agg(["mean", "std"])
      .round(2)
)

print(ship_mode_variability)
```

```python
region_analysis = (
    df.groupby("Region")["Shipping Lead Time"]
      .agg(["count", "mean", "std", "min", "max"])
      .round(2)
)

print(region_analysis)
```

```python
state_analysis = (
    df.groupby("State/Province")["Shipping Lead Time"]
      .agg(["count", "mean", "std", "min", "max"])
      .round(2)
)

print(state_analysis)
```

```python
top_volume_states = (
    state_analysis
    .sort_values("count", ascending=False)
    .head(10)
)

print(top_volume_states)
```

```python
overall_mean = df["Shipping Lead Time"].mean()

print("Overall Average Lead Time:", round(overall_mean, 2))
```

```python
state_analysis["Difference from Overall"] = (
    state_analysis["mean"] - overall_mean
).round(2)

print(
    state_analysis[
        ["count", "mean", "Difference from Overall"]
    ].sort_values("Difference from Overall", ascending=False).head(15)
)
```

```python
high_volume_bottlenecks = (
    state_analysis[
        (state_analysis["count"] >= 100) &
        (state_analysis["Difference from Overall"] > 0)
    ]
    .sort_values("Difference from Overall", ascending=False)
)

print(high_volume_bottlenecks[
    ["count", "mean", "Difference from Overall"]
])
```

```python
### Factory Route Analysis — Data Limitation

The project methodology requires Factory → Customer Region
and Factory → Customer State route analysis.

However, the provided dataset does not contain a Factory or
Factory Location column. Therefore, factory-based route
analysis cannot be performed using the current dataset.

The analysis will instead use the available Region and
State/Province fields for geographic shipping performance.
```

```python
state_summary = (
    df.groupby("State/Province")["Shipping Lead Time"]
      .agg(
          Shipment_Count="count",
          Average_Lead_Time="mean",
          Lead_Time_STD="std"
      )
      .round(2)
      .reset_index()
)

print(state_summary.head(20))
```

```python
df.to_csv("feature_engineered_data.csv", index=False)
```

```python
print("Saved:", "feature_engineered_data.csv")
```

```python
overall_mean = df["Shipping Lead Time"].mean()

print("Overall Average Lead Time:", round(overall_mean, 2))
```

```python
state_summary["Difference_from_Overall"] = (
    state_summary["Average_Lead_Time"] - overall_mean
).round(2)

print(
    state_summary[
        ["State/Province",
         "Shipment_Count",
         "Average_Lead_Time",
         "Difference_from_Overall"]
    ].head(20)
)
```

```python
potential_bottlenecks = (
    state_summary[
        (state_summary["Shipment_Count"] >= 100) &
        (state_summary["Difference_from_Overall"] > 0)
    ]
    .sort_values("Difference_from_Overall", ascending=False)
)

print(
    potential_bottlenecks[
        ["State/Province",
         "Shipment_Count",
         "Average_Lead_Time",
         "Difference_from_Overall"]
    ]
)
```

```python
state_summary["Performance_vs_Overall"] = (
    state_summary["Difference_from_Overall"]
    .apply(lambda x: "Above Overall" if x > 0 else "Below Overall")
)

print(
    state_summary[
        ["State/Province",
         "Shipment_Count",
         "Average_Lead_Time",
         "Difference_from_Overall",
         "Performance_vs_Overall"]
    ].head(20)
)
```

```python
state_summary["Performance_vs_Overall"] = (
    state_summary["Difference_from_Overall"]
    .apply(lambda x: "Above Overall" if x > 0 else "Below Overall")
)
```

```python
df = df.merge(
    state_summary[
        [
            "State/Province",
            "Shipment_Count",
            "Average_Lead_Time",
            "Lead_Time_STD",
            "Difference_from_Overall",
            "Performance_vs_Overall"
        ]
    ],
    on="State/Province",
    how="left"
)
```

```python
### Shipping Lead Time Validation

Shipping Lead Time was calculated as:

Ship Date - Order Date

The calculation was successfully validated. However, the
resulting lead times range from 904 to 1642 days because
the source data contains shipment dates substantially later
than the corresponding order dates.

This appears to be a source-data/date issue rather than a
calculation error. Therefore, the values are retained without
manual modification pending confirmation of the source data.
```

```python
print(df["Order Date"].dt.year.value_counts())
print()
print(df["Ship Date"].dt.year.value_counts())
```

```python
### Date / Shipping Lead Time Limitation

Shipping Lead Time was calculated using:

Ship Date - Order Date

The calculation was validated successfully. However, the
provided dataset contains Order Dates from 2024–2025 and
Ship Dates from 2026–2030, resulting in lead times ranging
from 904 to 1642 days.

Therefore, the calculated lead times are mathematically
correct based on the supplied dates, but they may not
represent realistic operational shipping durations.

No source dates were manually modified. The issue is
documented as a source-data limitation pending clarification.
```
